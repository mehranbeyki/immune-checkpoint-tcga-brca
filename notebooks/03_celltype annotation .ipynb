```r
library(Seurat)
library(ggplot2)
library(dplyr)
```

# Read RDS

```r
merged_sct <- readRDS("/Users/mehranbeyki/portfolio-projects/scrna-immune-checkpoint-response/data/processed/merged_integrated_clustered.rds")
```

# Markers

```r
# markers for cell type annotation
# cell type markers for PBMCs (peripheral blood mononuclear cells)
pbmc_markers <- c(
  "CD3D", "CD3E", "CD2",           # T-cell (general)
  "CD4", "IL7R",                   # CD4+ T-cell
  "CD8A", "CD8B",                  # CD8+ T-cell
  "FOXP3", "IL2RA",                # Treg
  "PDCD1", "CTLA4", "LAG3", "HAVCR2",  # T-cell (Exhaustion) 
  "NCAM1", "GNLY", "NKG7", "KLRD1",    # NK
  "CD14", "LYZ", "FCGR3A", "MS4A7",    # Monocyte
  "MS4A1", "CD19", "CD79A",            # B-cell
  "IGHG1", "MZB1", "JCHAIN",           # Plasma cell
  "FCER1A", "CST3", "CLEC9A",          # DC
  "PPBP", "PF4",                       # Platelet
  "HBB", "HBA1"                        # Red blood cell (likely contamination/purity issue)
)

# Filter the markers to only include those present in the dataset
pbmc_markers <- intersect(pbmc_markers, rownames(merged_sct))
```

## Mean expression + percentage of expressing cells in each cluster

```r
# measure the expression of the markers in each cluster
DotPlot(merged_sct, features = pbmc_markers, group.by = "seurat_clusters") +
  RotatedAxis() +
  theme(axis.text.x = element_text(size = 8))
ggsave("/Users/mehranbeyki/portfolio-projects/scrna-immune-checkpoint-response/results/figures/dotplot_pbmc_markers.png", width = 16, height = 9, dpi = 300)
```

```r
merged_sct <- PrepSCTFindMarkers(merged_sct)  # Prepares the SCT assay for marker detection
 
all_markers <- FindAllMarkers(
  merged_sct,
  assay = "SCT",
  only.pos = TRUE,
  min.pct = 0.25,
  logfc.threshold = 0.5
)
 
top5_per_cluster <- all_markers %>%
  group_by(cluster) %>%
  slice_max(order_by = avg_log2FC, n = 5)
 
write.csv(all_markers, "/Users/mehranbeyki/portfolio-projects/scrna-immune-checkpoint-response/results/tables/all_cluster_markers.csv", row.names = FALSE)
write.csv(top5_per_cluster, "/Users/mehranbeyki/portfolio-projects/scrna-immune-checkpoint-response/results/tables/top5_markers_per_cluster.csv", row.names = FALSE)
 
print(top5_per_cluster, n = 200)
 
saveRDS(merged_sct, "/Users/mehranbeyki/portfolio-projects/scrna-immune-checkpoint-response/data/merged_integrated_clustered_prepped.rds")
```
